# Converting numbers to 2^x

```python
def getNumData(num):
	output = {
		"original" : num,
		"positive" : num > 0,
		"integer" : bin(int(abs(num)))[2:],
		"decimal" : 0,
		"exponents" : [],
	}
	k = []
	d=abs(num)
	
	# 소수점 2진법으로 변환
	last = d-int(d)
	for i in range(1,10):
		tmp = last - 2**(-i)
		if tmp >= -0.0000001:
			last = tmp
			k += '1'
		else:
			k += '0'

	# 더러우니까 뒤에 0 잘라주기
	lastPos = 0
	for i, x in enumerate(k):
		if x == '1':
			lastPos = i
	output['decimal'] = ''.join(k)[:lastPos+1]


	intLen = len(output['integer'])
	for i, x in enumerate(output['integer']):
		if x == "1":
			output["exponents"].append(intLen-i-1)
	for i, x in enumerate(output["decimal"]):
		if x == "1":
			output["exponents"].append(-i-1)
	return output

print(getNumData(25))
print(getNumData(-3))
print(getNumData(13.625))
print(getNumData(-3.3))
```

```python
data = getNumData(12.125)
print(data)
```

```python
data["exponents"]
```

```python
def convertIt(datas):
	output = []
	for x in datas:
		if abs(x) > 1:
			output.append(convertIt(getNumData(x)["exponents"]))
		else:
			output.append(x)
	return output

exps = data["exponents"]
convertedExps = convertIt(exps)
```

```python
def convertIt(datas):
	output = []
	for x in datas:
		if abs(x) > 1:
			d = []
			for k in convertIt(getNumData(x)["exponents"]):
				d.append(f"2^{k}")
			output.append(' '.join(d))
		else:
			output.append(f"2^{abs(x)}")
	return output

exps = data["exponents"]
convertedTexts = ' '.join(convertIt(exps))
print(convertedTexts)
print([x >= 0 for x in data["exponents"]])
```

```python
checkPos = 0
for i, x in enumerate(data["exponents"]):
	if x < 0:
		checkPos = i
		break
print(data["exponents"], checkPos)
```

```python
targetText = convertIt([data["exponents"][checkPos]])[0].split(' ')[0]
print(targetText)
```

```python
isPositive = True
for x in convertedTexts.split(' '):
	if x == targetText:
		isPositive = False
	print(x, isPositive)
```
