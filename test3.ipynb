```python
def reset_var(targetText : str):
	global varNum, varText
	varNum = 0
	varText = targetText

reset_var('t')

import math

def get_var():
	global varNum
	varNum += 1
	mapSize = int(math.sqrt(varNum))+1
	x,y=0,0
	d=[[False for __ in range(mapSize)] for _ in range(mapSize)]

	boxSize = 0
	for _ in range(varNum):
		d[y][x]=True
		if x == 0 and y == boxSize:
			boxSize += 1
			x=boxSize
			y=0
		elif y < boxSize:
			y += 1
		elif y == boxSize:
			x -= 1

	output = ""
	for k in d:
		cnt = sum(k)
		tmp = ""
		if cnt > 0:
			tmp = varText
			for j in range(cnt-1):
				tmp = varText+"_{"+tmp+"}"
		output += tmp
	return varText+"_{"+output+"}"

for i in range(10):
	print(get_var())
```
