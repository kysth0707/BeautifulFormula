# Functions

```python
def reset_var():
	global varNum, varText
	varNum = 0
	varText = 't'

def get_var():
	global varNum
	varNum += 1
	return varText + "_{" + varText*varNum + "}"

reset_var()
```

```python
for i in range(3):
	print(get_var())
```

$$t_{t}$$
$$t_{tt}$$
$$t_{ttt}$$

```python
def get_1():
	return r"\frac{t}{t}"
print(get_1())
```

$$\frac{t}{t}$$

```python
reset_var()
def get_2():
	return r"\int_{-\frac{t}{t}}^{\frac{t}{t}}d"+get_var()
print(get_2())
```

$$\int_{-\frac{t}{t}}^{\frac{t}{t}}dt_{t}$$

```python
reset_var()
def get_3():
	v = get_var()
	return r"\sum_{"+v+r"="+get_1()+"}^{"+get_2()+"}"+v
print(get_3())
```

$$\sum_{t_{t}=\frac{t}{t}}^{\int_{-\frac{t}{t}}^{\frac{t}{t}}dt_{tt}}t_{t}$$

```python
reset_var()
def get_4():
	return r"\int_{-"+get_1()+r"}^{"+get_1()+r"}\int_{-"+get_1()+r"}^{"+get_1()+r"}"+f"d{get_var()}d{get_var()}"
print(get_4())
```

$$\int_{-\frac{t}{t}}^{\frac{t}{t}}\int_{-\frac{t}{t}}^{\frac{t}{t}}dt_{t}dt_{tt}$$

```python
reset_var()
def get_x_factorial(x):
	x=str(x)
	v = get_var()
	return r"\prod_{"+v+r"=\frac{t}{t}}^{"+x+"}"+v
print(get_x_factorial(1))
```

$$\prod_{t_{t}=\frac{t}{t}}^{1}t_{t}$$

```python
reset_var()
def get_x_plus_one(x):
	x=str(x)
	v = get_var()
	return r"\sum_{"+v+"="+get_1()+r"}^{"+x+"}"+v+r"\frac{"+get_2()+r"}{"+x+r"}"
print(get_x_plus_one("x"))
```

$$\sum_{t_{t}=\frac{t}{t}}^{x}t_{t}\frac{\int_{-\frac{t}{t}}^{\frac{t}{t}}dt_{tt}}{x}$$

```python
reset_var()
def get_ln_x(x):
	x=str(x)
	v = get_var()
	return r"\int_{"+get_1()+"}^{"+x+r"}"+v+r"^{-"+get_1()+"}d"+v
print(get_ln_x("x"))
```

$$\int_{\frac{t}{t}}^{x}t_{t}^{-\frac{t}{t}}dt_{t}$$

```python
reset_var()
def get_0():
	return r"\int_{"+get_1()+"}^{"+get_1()+"}d"+get_var()
print(get_0())
```

$$\int_{\frac{t}{t}}^{\frac{t}{t}}dt_{t}$$

```python
reset_var()
def get_e_x(x):
	x=str(x)
	v=get_var()
	return r"\sum_{"+v+r"="+get_0()+r"}^{"+get_4()+r"}\frac{"+x+r"^{"+v+r"}}{"+v+"!}"
print(get_e_x("x"))
```

$$\sum_{t_{t}=\int_{\frac{t}{t}}^{\frac{t}{t}}dt_{tt}}^{\int_{-\frac{t}{t}}^{\frac{t}{t}}\int_{-\frac{t}{t}}^{\frac{t}{t}}dt_{ttt}dt_{tttt}}\frac{x^{t_{t}}}{t_{t}!}$$

```python
reset_var()
def get_plus(d : list):
	output = ""
	for x in d:
		output += get_e_x(x)
	return get_ln_x(output)

print(get_plus(["x","y","z"]))
```

$$\int_{\frac{t}{t}}^{\sum_{t_{t}=\int_{\frac{t}{t}}^{\frac{t}{t}}dt_{tt}}^{\int_{-\frac{t}{t}}^{\frac{t}{t}}\int_{-\frac{t}{t}}^{\frac{t}{t}}dt_{ttt}dt_{tttt}}\frac{x^{t_{t}}}{t_{t}!}\sum_{t_{ttttt}=\int_{\frac{t}{t}}^{\frac{t}{t}}dt_{tttttt}}^{\int_{-\frac{t}{t}}^{\frac{t}{t}}\int_{-\frac{t}{t}}^{\frac{t}{t}}dt_{ttttttt}dt_{tttttttt}}\frac{y^{t_{ttttt}}}{t_{ttttt}!}\sum_{t_{ttttttttt}=\int_{\frac{t}{t}}^{\frac{t}{t}}dt_{tttttttttt}}^{\int_{-\frac{t}{t}}^{\frac{t}{t}}\int_{-\frac{t}{t}}^{\frac{t}{t}}dt_{ttttttttttt}dt_{tttttttttttt}}\frac{z^{t_{ttttttttt}}}{t_{ttttttttt}!}}t_{ttttttttttttt}^{-\frac{t}{t}}dt_{ttttttttttttt}$$

```python
reset_var()
def get_minus(a,b):
	return get_ln_x(r"\frac{"+get_e_x(a)+"}{"+get_e_x(b)+"}")
print(get_minus("x","y"))
```
